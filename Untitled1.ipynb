ici je voudrai savoir le nombre des personnes vieux et enfant qui on survecue

```python
age_survive = df[(df['Age'] > 60) & (df['Survived'] == 1 )]
age_total = df[df['Age'] > 60 ]['Survived']
enfant_total = df[df['Age'] <14 ]['Survived']
enfant_survive =df[(df['Age'] <14) & (df['Survived'] == 1 )]
```

```python
a=age_survive['Survived'].count()
b=age_total.count()
c=enfant_total.count()
d=enfant_survive['Survived'].count()
print('les personne ages qui etaient present ' + str(b) +' et seul '+ str(a) +' qui sont survive , ainsi que pour les enfant present sont '+ str(c) +' et seul '+ str(d) +' qui ont survive')
```

maintenat je veux visualiser cet statistique

```python
survivants_par_age = df[(df['Survived'] == 1) &  (df['Age']>60)].groupby('Age').size()

survivants_par_age.plot(kind='bar')
plt.xlabel('Age')
plt.ylabel('Nombre de survivants')
plt.title('Nombre de survivants selon l’âge')
plt.show()
```

```python
survivants_par_age = df[(df['Survived'] == 1) &  (df['Age']<14)].groupby('Age').size()

survivants_par_age.plot(kind='bar')
plt.xlabel('Age')
plt.ylabel('Nombre de survivants')
plt.title('Nombre de survivants selon l’âge')
plt.show()
```

```python
family = df.groupby('Name_matching')['Survived'].agg(
    total='count',
    survived='sum'
)

family['unsurvived'] = family['total'] - family['survived']

print(family)
```

```python
family = df.groupby('Name_matching')['Survived'].agg(
    total='count',
    survived='sum'
)

family['unsurvived'] = family['total'] - family['survived']

family = family[family['total'] >= 2]

family.sort_values(
    'total', ascending=False
).head(15)[['survived', 'unsurvived']].plot(kind='bar')

plt.xlabel('Famille')
plt.ylabel('Nombre de personnes')
plt.title('Survivants et morts dans les familles')
plt.xticks(rotation=45)
plt.legend(['Survivants', 'Morts'])
plt.tight_layout()
plt.show()
```
